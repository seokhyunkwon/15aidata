```python
# 22.10
f, s = map(int, input().split())
result=[]
if f>20 or f<1:
    print('잘못된 입력')
elif s>30 or s<10:
    print('잘못된 입력')
elif f>s:
    print('두 번째 값이 더 커야합니다.')
else:
    for i in range(f, s+1):
        result.append(2**i)
result.pop(1)
result.pop(-2)
print(result)
```
