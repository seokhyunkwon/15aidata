```python
# 문제. 1

# 가장 높은 점수, 낮은 점수, 총점, 평균, 분산을 구하는 프로그램을 만들어주세요 

# 입력예시 
# 70 80 85 38 43 30
# 또는 
# 100 18 38 94 93 12 39

# 공백으로 구분되어 몇개가 될지 모름

# 요구사항 
# 1. 함수이름은 스스로 지정해 주세요 
# 2. 가변 인수를 사용해 주세요 
def ksh(*args):
    max_s = max(args)
    min_s = min(args)
    total = sum(args)
    avg = total / len(args)
    var_total = 0
    for score in args:
        var_total += (score - avg) ** 2

    variance = var_total / len(args)
    
    return max_s, min_s, total, avg, variance

scores = list(map(int, input().split()))
max_s, min_s, total, avg, var = ksh(*scores)
print('최고점수:', max_s)
print('최저점수:',min_s)
print('총점:',total)
print(f"평균: {avg:.2f}")
print(f"분산: {var:.2f}")
```

```python
# 문제. 2

# prices = {
#     "아메리카노": 2500,
#     "카페라떼": 3500,
#     "바닐라라떼": 4000,
#     "초코케이크": 5500,
#     "샌드위치": 6000
# }

# 함수를 호출하면
# order(아메리카노=2, 카페라떼=1)
# order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
# order(아메리카노=1, 샌드위치=2)

# 실행결과
# 아메리카노 2개: 5000원
# 카페라떼 1개: 3500원
# 총 결제 금액: 8500원
# --------------------
# 바닐라라떼 2개: 8000원
# 초코케이크 1개: 5500원
# 샌드위치 1개: 6000원
# 총 결제 금액: 19500원
# --------------------
# 아메리카노 1개: 2500원
# 등록되지 않은 상품입니다: 허니브레드
# 총 결제 금액: 2500원

# 요구사항 
# 1. 함수 이름은 order로 만드세요.
# 2. 함수는 **kwargs를 사용해야 합니다.
# 3. kwargs의 key는 상품명, value는 주문 수량입니다.

prices = {
    "아메리카노": 2500,
    "카페라떼": 3500,
    "바닐라라떼": 4000,
    "초코케이크": 5500,
    "샌드위치": 6000
}

def order(**kwargs):
    total = 0
    for menu in kwargs:
        count = kwargs[menu]

        if menu in prices:
            price = prices[menu] * count
            total += price
            print(f'{menu} {count}개: {price}원')
        else:
            print('등록되지 않은 상품입니다:', menu)

    print(f'총 결제 금액: {total}원')
    print('--------------------')

order(아메리카노=2, 카페라떼=1)
order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
order(아메리카노=1, 허니브레드=1)
```

```python
# 문제 1. 재귀 호출로 합계와 계산 과정 출력하기

# 정수 n을 입력받아 n부터 1까지 더한 결과를 출력하는 프로그램을 만드세요.

# 단, 반드시 재귀 호출을 사용해야 하며, 계산 과정도 함께 출력해야 합니다.

# 입력:
# 5

# 출력: 
# 5 + 4 + 3 + 2 + 1 = 15

# 요구사항: 
# 1. 함수 이름은 recursive_sum으로 만드세요.
# 2. 반복문 for, while을 사용하지 마세요.
# 3. 재귀 호출을 사용하세요.
# 4. n == 1일 때 재귀 호출이 끝나도록 하세요.
# 5. 함수는 계산 과정 문자열과 합계를 함께 반환해야 합니다.
# 6. 최종 출력은 계산과정 = 결과 형태로 출력하세요.
# ---
def recursive_sum(n):
    if n == 1:
        return "1", 1
    next_num, result = recursive_sum(n - 1)
    return f'{n} + {next_num}' , n + result

n = int(input())
a, b = recursive_sum(n)
print(a,'=',b)
```

```python
# 문제 2. filter()와 lambda를 이용한 게시글 삭제 프로그램

# 다음과 같이 게시글 목록이 주어져 있습니다.

# posts = [
#     {"id": 1, "title": "파이썬 공부", "writer": "김철수"},
#     {"id": 2, "title": "오늘 할 일", "writer": "이영희"},
#     {"id": 3, "title": "장보기 목록", "writer": "박민수"},
#     {"id": 4, "title": "파이썬 복습", "writer": "김철수"},
#     {"id": 5, "title": "운동 계획", "writer": "최지수"}
# ]

# 삭제할 게시글 번호를 입력받아, 해당 id를 가진 게시글을 삭제한 뒤 남은 게시글 목록을 출력하는 프로그램을 작성하세요.

# 입력: 
# 3

# 출력: 
# 삭제 후 게시글 목록
# 1 파이썬 공부 김철수
# 2 오늘 할 일 이영희
# 4 파이썬 복습 김철수
# 5 운동 계획 최지수


# ---

posts = [
    {"id": 1, "title": "파이썬 공부", "writer": "김철수"},
    {"id": 2, "title": "오늘 할 일", "writer": "이영희"},
    {"id": 3, "title": "장보기 목록", "writer": "박민수"},
    {"id": 4, "title": "파이썬 복습", "writer": "김철수"},
    {"id": 5, "title": "운동 계획", "writer": "최지수"}
]

n = int(input())
posts = list(filter(lambda x: x['id'] != n, posts))

print('삭제 후 게시글 목록')
for post in posts:
    print(post['id'], post['title'], post['writer'])
```

```python
import random
names = [
    "권석현", "김규원", "김성준", "김태훈", "김형민", "문홍식", "박규민", "박성민", "박진영", "박찬현",
    "박희재", "손성욱", "신재민", "오은진", "원영배","이도영", "이수현", "이원석", "이진욱", "조현우","홍득훈"
]
student_data = {}
for name in names:
    student_data[name] = [random.randint(0, 100) for _ in range(10)]

class Student:
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def total(self):
        return sum(self.scores)
    
    def average(self):
        return self.total() / len(self.scores)
    
    def max_score(self):
        return max(self.scores)

    def min_score(self):
        return min(self.scores)
    
    def is_pass(self):
        if self.average() >= 70:
            return '합격'
        else:
            return '불합격'
        
    def print_info(self):
        print('이름: ', self.name)
        print('점수: ', self.scores)
        print('총점: ', self.total())
        print(f'평균: , {self.average():.2f}')
        print('최고점수: ', self.max_score())
        print('최저점수: ', self.min_score())
        print('결과: ', self.is_pass())
        print('-'* 50)

students = {}

for name in student_data:
    students[name] = Student(name, student_data[name])

for name in students:
    students[name].print_info()
```
