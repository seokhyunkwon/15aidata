```python
f, s = map(int, input().split(','))
for i in range(f, s+1):
    if i % 10 == 3 or i % 10 == 7:
        continue
    else:
        print(i)
```

```python
n = int(input())
for i in range(n):
    for j in range(i):
        print(' ', end='')
    for j in range(2*(n-i)-1):
        print('*', end='')
    print()

    
```

```python
f, s = map(int, input().split())
if f<s:
    print('잘못된 범위')
else:
    for i in range(f, s-1 , -1):
        if i % 3 == 0 and i % 7 ==0:
            print('FizzBuzz') 
        elif i % 3 == 0:
            print('Fizz')
        elif i % 7 == 0:
            print('Buzz')
        else:
            print(i)
```

```python
f, s = map(int, input().split())
result=[]

for i in range(f, s+1):
    result.append(round(i**0.5, 1))
print(result)
```

```python
f, s = map(int, input().split())
result = [round(i**0.5, 1) for i in range(f, s+1)]
print(result)
```
