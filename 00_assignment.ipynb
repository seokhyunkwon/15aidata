```python
# 문제. 1

# 가장 높은 점수, 낮은 점수, 총점, 평균, 분산을 구하는 프로그램을 만들어주세요 

# 입력예시 
# 70 80 85 38 43 30
# 또는 
# 100 18 38 94 93 12 39

# 공백으로 구분되어 몇개가 될지 모름

# 요구사항 
# 1. 함수이름은 스스로 지정해 주세요 
# 2. 가변 인수를 사용해 주세요 
def ksh(*args):
    max_s = max(args)
    min_s = min(args)
    total = sum(args)
    avg = total / len(args)
    var_total = 0
    for score in args:
        var_total += (score - avg) ** 2

    variance = var_total / len(args)
    
    return max_s, min_s, total, avg, variance

scores = list(map(int, input().split()))
max_s, min_s, total, avg, var = ksh(*scores)
print('최고점수:', max_s)
print('최저점수:',min_s)
print('총점:',total)
print(f"평균: {avg:.2f}")
print(f"분산: {var:.2f}")
```

```python
# 문제. 2

# prices = {
#     "아메리카노": 2500,
#     "카페라떼": 3500,
#     "바닐라라떼": 4000,
#     "초코케이크": 5500,
#     "샌드위치": 6000
# }

# 함수를 호출하면
# order(아메리카노=2, 카페라떼=1)
# order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
# order(아메리카노=1, 샌드위치=2)

# 실행결과
# 아메리카노 2개: 5000원
# 카페라떼 1개: 3500원
# 총 결제 금액: 8500원
# --------------------
# 바닐라라떼 2개: 8000원
# 초코케이크 1개: 5500원
# 샌드위치 1개: 6000원
# 총 결제 금액: 19500원
# --------------------
# 아메리카노 1개: 2500원
# 등록되지 않은 상품입니다: 허니브레드
# 총 결제 금액: 2500원

# 요구사항 
# 1. 함수 이름은 order로 만드세요.
# 2. 함수는 **kwargs를 사용해야 합니다.
# 3. kwargs의 key는 상품명, value는 주문 수량입니다.

prices = {
    "아메리카노": 2500,
    "카페라떼": 3500,
    "바닐라라떼": 4000,
    "초코케이크": 5500,
    "샌드위치": 6000
}

def order(**kwargs):
    total = 0
    for menu in kwargs:
        count = kwargs[menu]

        if menu in prices:
            price = prices[menu] * count
            total += price
            print(f'{menu} {count}개: {price}원')
        else:
            print('등록되지 않은 상품입니다:', menu)

    print(f'총 결제 금액: {total}원')
    print('--------------------')

order(아메리카노=2, 카페라떼=1)
order(바닐라라떼=2, 초코케이크=1, 샌드위치=1)
order(아메리카노=1, 허니브레드=1)
```
