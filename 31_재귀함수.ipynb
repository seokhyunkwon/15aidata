```python
def hello(count):
    if count == 0:
        return
    print('Hello, World!', count)

    count -= 1
    hello(count)

hello(5)
```

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)

print(factorial(5))
```

```python
# 31.5
def fib(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fib(n-2) + fib(n-1)

n = int(input())
print(fib(n))
```
