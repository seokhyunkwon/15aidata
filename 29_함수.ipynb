```python
# 29.8
x, y = map(int, input().split())
def calc(a, b):
    return a+b, a-b, a*b, float(a/b)
a, s, m, d = calc(x, y)
print(f'덧셈: {a}, 뺄셈: {s}, 곱셈: {m}, 나눗셈: {d:.2f}')
```
