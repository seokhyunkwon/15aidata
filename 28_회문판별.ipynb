```python
# 거꾸로 읽어도 같은 문자
word = input('단어입력: ')
is_palindrome = True
for i in range(len(word) // 2):
    if word[i] != word[-1 - i]:
        is_palindrome = False
        break

print(is_palindrome)
```

```python
# 시퀀스 뒤집기로 문자 검사
word = input('단어입력: ')
print(word == word[::-1])
```

```python
# 리스트와 reversed 사용
word = 'level'
list(word) == list(reversed(word))
```

```python
# join 메소드와 reversed 사용
word = 'level'
word == ''.join(reversed(word))
```

```python
# N-gram 만들기
text = 'Hello'
for i in range(len(text) - 1):
    print(text[i], text[i+1], sep='')
```

```python
# 반복문으로 N-gram 출력
text = 'this is python script'
words = text.split()

for i in range(len(words) - 1):
    print(words[i], words[i + 1])
```

```python
# zip 으로 2-gram 만들기
text = 'hello'

two_gram = zip(text, text[1:])
for i in two_gram:
    print(i[0], i[1], sep='')
```

```python
# zip 과 리스트 표현식으로 N-gram 만들기
text = 'hello'
[text[i:] for i in range(3)]
```

```python
list(zip(*[text[i:] for i in range(3)]))
```

```python
# 28.3 369p
n = int(input())
text = input()
words = text.split()
if len(words) < n:
    print('wrong')
else:
    n_gram = zip(*[words[i:] for i in range(n)])
    for i in n_gram:
        print(i)
```
