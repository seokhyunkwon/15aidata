```python
# 기반 클래스의 속성 사용하기
class Person:
    def __init__(self):
        print('Person __init__')
        self.hello = '안녕'

class Student(Person):
    def __init__(self):
        print('Student __init__')
        super().__init__()
        self.school = '파이썬 코딩 도장'

james = Student()
print(james.school)
print(james.hello)
```

```python
# 다중상속 사용

class Person:
    def greeting(self):
        print('안녕')

class University:
    def manage(self):
        print('학점 관리')

class Under(Person, University):
    def study(self):
        print('공부')

james = Under()
james.greeting()
james.manage()
james.study()
```

```python
# 추상 클래스 사용하기

from abc import *

class StudentBase(metaclass=ABCMeta):
    @abstractmethod
    def study(self):
        pass

    @abstractmethod
    def go_to_school(self):
        pass

class Student(StudentBase):
    def study(self):
        print('공부하기')

james = Student()
james.study()
```

```python
# 추상 클래스 사용하기

from abc import *

class StudentBase(metaclass=ABCMeta):
    @abstractmethod
    def study(self):
        pass

    @abstractmethod
    def go_to_school(self):
        pass

class Student(StudentBase):
    def study(self):
        print('공부하기')
    def go_to_school(self):
        print('학교가기')

james = Student()
james.study()
james.go_to_school()
```

```python
# 36.8 
class AdvancedList(list):
    def replace(self, old, new):
        for i in range(len(self)):
            if self[i] == old:
                self[i] = new

x = AdvancedList([1, 2, 3, 1,2,3,1,2,3])
x.replace(1, 100)
print(x)
```

```python
# 36.9

class Animal:
    def eat(self):
        print('먹다')

class Wing:
    def flap(self):
        print('파닥거리다')

class Bird(Animal, Wing):
    def fly(self):
        print('날다')

b = Bird()
b.eat()
b.flap()
b.fly()
print(issubclass(Bird, Animal))
print(issubclass(Bird, Wing))
```
