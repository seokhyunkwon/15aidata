```python
# 3의 배수일 때와 5의 배수일 때 처리하기
for i in range(1, 101):
    if i % 3 ==0:
        print('Fizz')
    elif i % 5 ==0:
        print('Buzz')
    else:
        print(i)
```

```python
# 3과 5의 공배수 처리
for i in range(1, 101):
    if i % 3 == 0 and i % 5 ==0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print('Fizz')
    elif i % 5 == 0:
        print('Buzz')
    else:
        print(i)
```

```python
# 논리 연산자 없이 3과 5의 공배수 ㅊ ㅓ리하기
for i in range(1, 101):
    if i % 15 == 0:
        print('FizzBuzz')
    elif i % 3 ==0:
        print('Fizz')
    elif i % 5 ==0:
        print('Buzz')
    else:
        print(i)
```

```python
# 코드 단축하기
for i in range(1,101):
    print('Fizz' * (i %  3 ==0) + 'Buzz' * (i % 5 ==0) or i)
```

```python
# 20.8
first, second = map(int, input().split())
while True:
    if first < 1 or first > 1000:
        print('잘못된 범위')
        break
    elif second < 10 or second > 1000:
        print('잘못된 범위')
        break
    elif first > second:
        print('두 번째 값이 더 커야합니다.')
        break
    else:
        for i in range(first, second+1):
            if i % 35 == 0:
                print('FizzBuzz')
            elif i % 5 == 0:
                print('Fizz')
            elif i % 7 == 0:
                print('Buzz')
            else:
                print(i)
        break
```
