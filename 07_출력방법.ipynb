```python
# 값을 여러 개 출력하기

print(1, 2, 3)
```

```python
# sep로 값 사이에 문자 넣기
print(1, 2, 3, sep = ", ")
print(2026, 1048, sep = " x ")
```

```python
# 줄바꿈 활용
print(1,2,3,sep = "\n")
print("1\n2\n3")
```

```python
# end 활용하기
print(1)
print(2)
print(3)
```

```python
print(1, end = "찍")
print(2, end = "찍")
print(3, end = "찍")
```

```python
year, month, day, hour, minute, second = input().split()
print(year, month, day, sep="-", end="T")
print(hour, minute, second, sep=":")
```
