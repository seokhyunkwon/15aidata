```python
# 위치 인수와 리스트 언패킹
def print_num(a,b,c):
    print(a)
    print(b)
    print(c)

print_num(4,5,6)
```

```python
x = [10, 20, 30]
print_num(*x)
```

```python
print_num(*[10,20,30])
```

```python
# 가변 인수 함수
def print_num(*args):
    for arg in args:
        print(arg)

print_num(165,156,1,1,1,1,1,1)
print()
print_num(165,156,1,1,1,1,1123,65,8,8,8,8,8)
```

```python
# 키워드 인수
def info(name, age, addr):
    print('이름: ', name)
    print('나이: ', age)
    print('주소: ', addr)

info(age=60, name='홍길동', addr='대구')    
```

```python
# 키워드 인수와 딕셔너리 언패킹 
def info(name, age, addr):
    print('이름: ', name)
    print('나이: ', age)
    print('주소: ', addr)

x = {'name':'홍길동', 'age':30, 'addr':'대구'}
info(**x)
```

```python
# 키워드 인수를 사용하는 가변 인수 함수
def info(**kwargs):
    for kw, arg in kwargs.items():
        print(kw, ': ', arg, sep='')
info(name='홍길동')
print()
info(name='홍길동', age=30, addr='대구', hb='코딩')
```

```python
x = {'name':'홍길동'}
info(**x)
```

```python
# 매개변수에 초깃값 지정하기
def iinfo(name, age, addr='비공개'):
    print('이름: ', name)
    print('나이: ', age)
    print('주소: ', addr)

iinfo('홍길동', 30)
```

```python
# 초깃값이 저장된 매개변수 위치는 마지막!!! 중간이면 안됨
iiinfo(name, addr='비공개', age)
```

```python
# 30.6 
kor, eng, math, snc = 100, 86, 81, 91
def get_max_score(*args):
    return max(args)
max_score = get_max_score(kor, eng, math, snc)
print("높은 점수:", max_score)
max_score = get_max_score(eng, snc)
print("높은 점수:", max_score)
```

```python
# 30.7
k, e, m, s = map(int, input().split())
def m_m_s(*args):
    return min(args), max(args)
def get_avg(**kwargs):
    return sum(kwargs.values()) / len(kwargs)

min_s, max_s = m_m_s(k, e, m, s)
avg_s = get_avg(k=k,e=e,m=m,s=s)
print(f'낮은 점수: {min_s:.2f}, 높은 점수: {max_s:.2f}, 평균 점수: {avg_s:.2f}')
min_s, max_s = m_m_s(e,s)
avg_s = get_avg(e=e,s=s)
print(f'낮은 점수: {min_s:.2f}, 높은 점수: {max_s:.2f}, 평균 점수: {avg_s:.2f}')
```
