```python
import math
class Point2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y
p1 = Point2D(30, 20)
p2 = Point2D(60, 50)
a = p2.x - p1.x
b = p2.y - p1.y

c = math.sqrt(a**2 + b**2)
d = math.sqrt(pow(a,2) + pow(b,2))
print(c)
print(d)
```

```python
# 37.2 
class R:
    def __init__(self,x1,y1,x2,y2):
        self.x1 = x1
        self.y1 = y1
        self.x2 = x2
        self.y2 = y2

rect = R(20,20,40,30)
w = abs(rect.x2 - rect.x1)
h = abs(rect.y2 - rect.y1)
area = w * h
print(area)
```

```python
# 37.3
import math

class P2:
    def __init__(self, x=0, y=0):
        self.x =x
        self.y = y

length = 0.0
p = [P2(), P2(), P2(), P2()]
p[0].x, p[0].y, p[1].x, p[1].y, p[2].x, p[2].y, p[3].x, p[3].y = map(int, input().split())
for i in range(len(p)-1):
    a = p[i+1].x - p[i].x
    b = p[i+1].y - p[i].y
    length += math.sqrt(pow(a,2) + pow(b,2))

print(length)
```
