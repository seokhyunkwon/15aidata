```python
y = [10,20,30]
try:
    index, x = map(int, input('인덱스와 나눌 숫자 입력').split())
    print(y[index] / x)
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다')
except IndexError:
    print('인덱스 범위를 벗어났습니다')
```

```python
try:
    x = int(input())
    y = 10 / x
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다')
else:
    print(y)
finally:
    print('코드 실행 완료')
```

```python
# 예외 발생시키기
try:
    x=int(input('3의배수'))
    if x % 3 != 0:
        raise Exception('3의 배수가 아님')
    print(x)
except Exception as e:
    print('예외 발생', e)
```

```python
# 현재 예외를 다시 발생시키기
def t():
    try:

        x = int(input('3배수'))
        if x % 3 != 0:
            raise Exception('3의 배수가 아님')
        print(x)
    except Exception as e:
        print('t함수에서 예외발생', e)
        raise    

try:
    t()
except Exception as e:
    print('스크립트 예외 발생', e)
```

```python
# 예외 만들기
class NotThree(Exception):
    def __init__(self):
        super().__init__('3의 배수가 아님')

def t():
    try:
        x = int(input('3배수'))
        if x % 3 != 0:
            raise NotThree()
        print(x)
    except Exception as e:
        print('예외발생', e)

t()
```

```python
# 38.7
class NotPalindromeError(Exception):
  def __init__(self):
    super().__init__('회문이 아닙니다.')

def palindrome(word):
  if word != word[::-1]:
    raise NotPalindromeError
  print(word)

try:
  word = input()
  palindrome(word) 
except NotPalindromeError as e:
  print(e)
```
