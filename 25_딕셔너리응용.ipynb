```python
# 표현식에서 if 사용하기
x = {'a':10, 'b':20, 'c':30, 'd':40}
for key, value in x.items():
    if value == 20:
        del x[key]


# 반복 중에 키값 쌍을 삭제하면 오류발생
        
print(x)
```

```python
# 딕셔너리의 할당과 복사
x = {'a':10, 'b':20, 'c':30, 'd':40}
y = x

print(y is x)
y['a'] = 99
print(x)
print(y)
```

```python
y=x.copy()
x is y
```

```python
# 중첩 딕셔너리의 할당과 복사
x = {'a':{'python':'2.7'}, 'b':{'python' : '3.6'}}
y = x.copy()
y['a']['python'] = '2.7.15'
print(x)
print(y)
```

```python
# 중첩 딕셔너리를 완전히 복사하려면 deepcopy를 써야함
import copy
x = {'a':{'python':'2.7'}, 'b':{'python' : '3.6'}}
y = copy.deepcopy(x)
y['a']['python'] = '2.7.15'
print(x)
print(y)
```
