```python
# 숫자와 영문 문자 조합
re.match('[a-zA-Z0-9]+', 'abc123')
```

```python
# 특정 문자 범위에 포함되지 않는지 판단
re.match('[^0-9]+', 'abc')
```

```python
re.match('[^A-Z]+', '123')
```

```python
# 특정 문자 범위가 아닌지 판단할 때는 [] 안에 ^ 사용, 특정 문자열로 시작하는 지 판단할 때는 앞에 ^ 사용, 특정 문자열로 끝나는 지 판단할 때는 뒤에 $ 사용
re.search('^[A-Z]', 'Hello')
```

```python
re.search('[0-9]+$', 'abcfweafwaef123')
```

```python
# 특수 문자 판단하기 \ 를 붙이기 [] 안에서는 안붙여도됨
re.search('\*+', '1 ** 2')
```

```python
re.match('[$()a-zA-Z0-9]+', '$(document)')
```

```python
# \d : [0-9]와 동일
# \D : [^0-9]와 동일 
# \s : 공백 문자, \S : 공백 문자가 아닌 것
# \w : [a-zA-Z0-9_]와 동일
# \W : [^a-zA-Z0-9_]와 동일
```

```python
re.match('\d+', '1234')
```

```python
re.match('\D+', 'abc')
```

```python
re.match('\w+', 'abc123')
```

```python
re.match('\W+', '!!@#!@#!@$!@$')
```

```python
# 공백처리 \s
re.match('[a-zA-Z0-9 ]+', ' abc 123  ')
```

```python
re.match('[a-zA-Z0-9\s]+', ' abc 123  ')
```

```python
# 그룹 사용하기
m = re.match('([0-9]+) ([0-9]+)', '10 2994')
m.group(1)
```

```python
m.group(2)
```

```python
m.group(0)
```

```python
m.group()
```

```python
m.groups()
```

```python
# (?P<그룹이름>정규표현식)
m = re.match('(?P<func>[a-zA-Z_][a-zA-Z0-9_]+)\((?P<arg>\w+)\)', 'print(1234)')
```

```python
m.group('func')
```

```python
m.group('arg')
```

```python
# 패턴에 매칭되는 모든 문자열 가져오기
re.findall('[0-9]+', '1 2 Fizz 4 Buzz Fizz 7 8')
```

```python
# 문자열 바꾸기
re.sub('apple|orange', 'fruit', 'apple box orange tree')
```

```python
re.sub('[0-9]+', 'n', '1 2 Fizz 4 Buzz Fizz 7 8')
```

```python
re.sub('[0-9]+', lambda x: str(int(x.group()) * 10), '1 2 Fizz 4 Buzz Fizz 7 8')
```

```python
# 찾은 문자열을 결과에 다시 사용
re.sub('([a-z]+) ([0-9]+)', '\\2 \\1 \\2 \\1', 'hello 1234')
```

```python
re.sub('({\s*})"(\w+)":\s*"(\w+)"(/s*})', '<\\2>\\3</\\2>', '{ "name" : "james" }')
##?!?!?!?!?!?
```
