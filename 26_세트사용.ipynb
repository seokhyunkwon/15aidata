```python
# 집합 연산 사용
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
#or
print(a|b)
print(set.union(a,b))
#and
print(a&b)
print(set.intersection(a,b))
# -
print(a-b)
print(set.difference(a,b))
# XOR
print(a^b)
print(set.symmetric_difference(a,b))
```

```python
# 집합 연산 후 할당 연산자 사용
a = {1, 2, 3, 4}
a |= {5}
print(a)
a = {1, 2, 3, 4}
a.update({5})
print(a)
```

```python
a = {1, 2, 3, 4}
a &= {0,2,3,4}
print(a)
a = {1, 2, 3, 4}
a.intersection_update({0,2,3,4})
print(a)
```

```python
a = {1, 2, 3, 4}
a-= {3}
print(a)
a = {1, 2, 3, 4}
a.difference_update({3})
print(a)
```

```python
a = {1, 2, 3, 4}
a ^= {3,4,5,6}
print(a)
a = {1, 2, 3, 4}
a.symmetric_difference_update({3,4,5,6})
print(a)
```
