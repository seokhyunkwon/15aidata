```python
# 이터레이터 만들기
class Counter:
    def __init__(self, stop):
        self.current = 0
        self.stop = stop

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < self.stop:
            r = self.current
            self.current += 1
            return r
        else:
            raise StopIteration
for i in Counter(3):
    print(i, end= ' ')
```

```python
# 이터레이터 언패킹
a, b, c = Counter(3)
print(a, b, c)
```

```python
# 인덱스로 접근할 수 있는 이터레이터
class Counter:
    def __init__(self, stop):
        self.stop = stop
    def __getitem__(self, index):
        if index < self.stop:
            return index
        else:
            raise IndexError

print(Counter(3)[0], Counter(3)[1], Counter(3)[2])
for i in Counter(3):
    print(i, end=' ') 
```

```python
# 39.6 
class M:
    def __init__(self, stop, m):
        self.stop = stop
        self.m = m
        self.current = 0
    def __iter__(self):
        return self
    def __next__(self):
        self.current += 1
        if self.current < self.stop:
            return self.current * self.m
        else:
            raise StopIteration
        
for i in M(20, 3):
    print(i, end=' ')
print()
for i in M(30, 5):
    print(i, end=' ')
        
```
