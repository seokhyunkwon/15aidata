```python
# 동작 방식
if True:
    print('참')
else:
    print('거짓')

if None:
    print('참')
else:
    print('거짓')
```

```python
# 문자열 지정하기
## 빈문자열은 거짓, 문자열은 참
```

```python
# 조건식 여러 개 지정
x=10
y=20
if x == 10 and y == 20:
    print('참')
else:
    print('거짓')
```

```python
# 중첩 if 조건문과 논리 연산자
x=10
if x>0:
    if x<20:
        print("x는 20보다 작은 양수 입니다")
```

```python
if x>0 and x<20:
    print('x는 20보다 작은 양수입니다.')
```

```python
if 0<x<20:
    print('x는 20보다 작은 양수입니다.')
```
