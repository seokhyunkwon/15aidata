```python
# 2차원 리스트

a=[[10,20], [30,40], [50,60]]
a[0][0]
```

```python
a=[[10,20], [30,40], [50,60]]
for x, y in a:
    print(x, y)
```

```python
a=[[10,20], [30,40], [50,60]]
for i in a:
    for j in i:
        print(j, end=' ')
    print()
```

```python
a=[[10,20], [30,40], [50,60]]
for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end=' ')
    print()
```

```python
a=[[10,20], [30,40], [50,60]]

i=0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
a=[[10,20], [30,40], [50,60]]
i=0
while i < len(a):
    j=0
    while j < len(a[i]):
        print(a[i][j], end=' ')
        j +=1
    print()
    i += 1
```

```python
# 반복문으로 리스트 만들기
a=[]
for i in range(3):
    line=[]
    for j in range(2):
        line.append(0)
    a.append(line)

print(a)
```

```python
a = [[0 for j in range(2)] for i in range(3)]
print(a)
```

```python
a = [3, 1, 2, 4, 5]
b = []

for i in a:
    line=[]
    for j in range(i):
        line.append(0)
    b.append(line)

print(b)
```

```python
# 할당과 복사
a=[[10,20], [30,40]]
b = a
b[0][0] = 500
print(a)
print(b)
```

```python
a=[[10,20], [30,40]]
import copy
b = copy.deepcopy(a)
b[0][0] = 500
print(a)
print(b)
```
