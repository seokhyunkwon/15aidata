```python
# nonlocal 지역변수로 바깥쪽 함수의 지역변수 변경가능
def A():
    x=10
    def B():
        nonlocal x
        x=20

    B()
    print(x)

A()
```

```python
# 가장 가까운 바깥쪽 지역변수 사용 x는 B() 꺼 y는 A() 꺼 사용
def A():
    x=10
    y=100
    def B():
        x=20
        def C():
            nonlocal x
            nonlocal y
            x += 30
            y += 300
            print(x)
            print(y)
        C()
    B()

A()    
```

```python
# global로 전역 변수 사용
x=1
def A():
    x=10
    def B():
        x=20
        def C():
            global x
            x += 30
            print(x)
        C()
    B()

A()
```

```python
# 클로저 사용하기
def calc():
    a=3
    b=5
    def mul_add(x):
        return a*x + b
    return mul_add

c = calc()
print(c(1), c(2), c(3), c(4), c(5))
```

```python
# lambda로 클로저 만들기
def calc():
    a=3
    b=5
    return lambda x: a* x + b

c = calc()
print(c(1), c(2), c(3), c(4), c(5))
```

```python
# 클로저의 지역변수 변경
def calc():
    a=3
    b=5
    total = 0
    def mul_add(x):
        nonlocal total
        total = total + a*x + b
        print(total)
    return mul_add

c = calc()
c(1)
c(2)
c(3)
```

```python
# 33.6
def countdown(n):
    def count():
        nonlocal n
        r = n
        n -= 1
        return r
    return count

n = int(input())
c = countdown(n)
for i in range(n):
    print(c(), end=' ')
```
