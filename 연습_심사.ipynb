```python
import copy
col, row = map(int, input().split())
matrix = []
for i in range(row):
    matrix.append(list(input()))
mx = copy.deepcopy(matrix) 
for i in range(row):
    for j in range(col):
        if matrix[i][j] == '*':
            continue 
            
        else:
            mx[i][j] = 0
            for y in range(i - 1, i + 2): 
                for x in range(j -1, j + 2): 
                    if y < 0 or x < 0 or y >= row or x >= col:
                        continue
                    elif matrix[y][x] == '*':
                        mx[i][j] += 1

for i in range(row):
    for j in range(col):
        print(mx[i][j], end='')
    print()
```

```python
# 31.4
def is_p(word):
    if len(word) < 2:
        return True
    if word[0] != word[-1]:
        return False
    return is_p(word[1:-1])

print(is_p('hello'))
print(is_p('level'))
```

```python
# 31.5
def fib(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fib(n-2) + fib(n-1)

n = int(input())
print(fib(n))
```
