```python
# for
for i in range(100):
    print('Hello, World!')
```

```python
for i in range(100):
    print('Hello, World!', i)
```

```python
# range와 for 응용
for i in range(5, 12):
    print('Hello ,World!', i)
```

```python
# 증가폭 사용
for i in range(0,10,2):
    print('Hello, World', i)
```

```python
# 감소시키기
for i in range(10, 0 , -1):
    print('Hello, World!', i)
```

```python
for i in reversed(range(10)):
    print('Hello, World!', i)
```

```python
# 입력한 횟수대로 반복하기

cnt = int(input('반복할 횟수: '))
for i in range(cnt):
    print('Hello, World!', i)
```

```python
# 시퀀스 객체 반복
a=[10, 20, 30, 40, 50]
for i in a:
    print(i)
```

```python
f=('apple', 'orange', 'grape')
for fruit in f:
    print(fruit)
```

```python
for letter in 'Python':
    print(letter, end=' ')
```

```python
for letter in reversed('Python'):
    print(letter, end=' ')
```
