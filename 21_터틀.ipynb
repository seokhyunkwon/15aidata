```python
import turtle as t

# 거북이 설정
t.bgcolor("white")
t.speed(0)
t.pensize(3)
t.color("red")

# 거북이가 그리면서 이동하는 "사랑해"
def write_message():
    """거북이가 이동하면서 메시지 그리기"""
    t.penup()
    t.goto(-100, 80)
    t.pendown()
    
    # 첫 번째: 사 (간단한 상자)
    for i in range(4):
        t.forward(40)
        t.right(90)
    
    # 이동
    t.penup()
    t.forward(60)
    t.pendown()
    
    # 두 번째: 랑 (삼각형)
    t.forward(40)
    t.right(120)
    t.forward(40)
    t.right(120)
    t.forward(40)
    
    # 이동
    t.penup()
    t.forward(60)
    t.pendown()
    
    # 세 번째: 해 (별)
    for i in range(5):
        t.forward(40)
        t.right(144)

write_message()

# 하트 그리기
t.penup()
t.goto(0, -100)
t.pendown()
t.color("pink")
t.pensize(4)

# 왼쪽 하트
t.circle(30, 180)
t.right(90)
t.forward(60)
t.right(90)

# 오른쪽 하트
t.circle(30, 180)

t.hideturtle()
t.exitonclick()
```
